```python
# Imports

import subprocess
import io
import numpy as np
import pandas as pd
```

```python
SUBJECTS = [
    "Sitophilus granarius",
    "Sitophilus oryzae",
    "Sitophilus zeamais",
    
    "Dendroctonus ponderosae",
    "Dendroctonus valens",
    "Ips typographus",
    "Orthorhinus cylindrirostris",
    "Xyleborus affinis",
    
    "Anthonomus grandis",
    "Elaeidobius kamerunicus",
    
    "Curculio glandium",
    "Hypothenemus hampei",
    
    "Diaprepes abbreviatus",
    "Rhynchophorus ferrugineus",

    "Tribolium castaneum",
    "Acyrthosiphon pisum"
]

print(len(SUBJECTS))
```

```python
ESEARCH_DTYPES = {
    "Run": str,
    "BioProject": str,
    "Sample": str,
    "BioSample": str,
    "ScientificName": str,
    
    "size_MB": int,
    "spots": int,
    "bases": int,
    
    "Experiment": str,
    "SRAStudy": str
}

def run_esearch(species_name: str) -> pd.DataFrame:
    query = f'\\"{species_name}\\"[Organism] AND WGS[Strategy] AND ILLUMINA[Platform] AND PAIRED[Layout]'
    cmd = f'esearch -db sra -query "{query}" | efetch -format runinfo'

    # print("$", cmd)
    result = subprocess.run(
        cmd, 
        shell=True, 
        capture_output=True, 
        text=True, 
        check=True
    )

    if not result.stdout:
        return pd.DataFrame()

    df = pd.read_csv(
        io.StringIO(result.stdout),
        dtype=ESEARCH_DTYPES,
        usecols=ESEARCH_DTYPES.keys()
    )
    
    return df
```

```python
for subject in SUBJECTS:
    df = run_esearch(subject)
    print(subject, "-", len(df))
```
